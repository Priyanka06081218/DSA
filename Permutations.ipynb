```python
def permute(nums):

    res = []

    def backtrack(i):

        if i == len(nums):
            res.append(nums[:])
            return

        for j in range(i, len(nums)):
            nums[i], nums[j] = nums[j], nums[i]
            backtrack(i + 1)
            nums[i], nums[j] = nums[j], nums[i]

    backtrack(0)
    return res


nums = [1, 2, 3]

print(permute(nums))
```
