```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        max_len = 0
        for start in range(len(s)):
            freq = {}
            total = 0
            for i in s[start:]:
                freq[i] = freq.get(i, 0) + 1
                if freq[i] > 1:
                    break
                total += 1

            max_len = max(max_len, total)

        return max_len

# Single Example Execution
sol = Solution()
output = sol.lengthOfLongestSubstring("abcabcbb")

print("Result:", output)
```
