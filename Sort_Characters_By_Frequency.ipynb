```python
from typing import List
class Solution:
    def frequencySort(self, s: str) -> str:
        freq = {}
        for char in s:
            freq[char] = freq.get(char, 0) + 1
            sorted_chars = sorted(
            freq.items(),
            key=lambda x: x[1],
            reverse=True
            )
            res = ""
            for char, count in sorted_chars:
                res += char * count
        return res

solution = Solution()

s = "tree"

result = solution.frequencySort(s)

print("Input:", s)
print("Output:", result)
```
