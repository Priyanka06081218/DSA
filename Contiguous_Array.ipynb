```python
from typing import List

class Solution:
    def findMaxLength(self, nums: List[int]) -> int:
        res = 0
        count = 0

        seen = {0: -1}

        for i in range(len(nums)):
            if nums[i] == 0:
                count -= 1
            else:
                count += 1

            if count in seen:
                res = max(res, i - seen[count])
            else:
                seen[count] = i

        return res


sol = Solution()
output = sol.findMaxLength([0, 1, 0, 1])

print("Result:", output)
```
